```python
# Item 6: Mutiple Assignments

snacks = [("bacon", 350), ("donut", 240), ("muffin", 190)]

# start=1; the index value from which the counter is to be started, by default it is 0 
for rank, (name, calories) in enumerate(snacks, 1):
    print(f'#{rank}: {name}: {calories}')
```

```python
# Item 8: zip

names = ["Andreas", "Roessler", "HS", "Esslingen"]
no_of_chars = [len(n) for n in names]
for name, noc in zip(names, no_of_chars):
    print(f"{name} has {noc} chars")
```

```python
# Item 10: Walrus: that assigns values to variables as part of a larger expression
a = range(11)
if (n := len(a)) > 10:
    print(f"List is too long ({n} elements, expected <= 10)")
```

```python
# Item 13: Unpacking

first, *others, last = a
print(first, last)

b = ["Hallo", "Welt"]
first, *others, last = b
print(first, last)

b = ["Hallo"]
# first, *others, last = b              # ERROR not enough values to unpack 
print(first, last)

```

```python
# Item 16: Handle Missing Dict Keys
counters = {
    "students": 0
}
key = "employees"
print(counters)
```

```python
counters[key] = counters.get(key, 0) + 1
print(counters)
```
