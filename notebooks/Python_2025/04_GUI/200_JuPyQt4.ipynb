# Einführung in PyQt

Von PyQt benötigen wir zunächst einige Klassen. Damit PyQt in Jupyter gestartet werden kann, wird eine spezielle EventLoop benötigt.


```python
import sys
from PyQt6.QtWidgets import QApplication, QWidget, QVBoxLayout, QPushButton, QLabel
from PyQt6.QtCore import QTimer

# Für Qt in Jupyter
import asyncio
from qasync import QEventLoop, asyncSlot
```

```python
# Eine einfache PyQt5-Demo
class SimpleApp(QWidget):
    def __init__(self):
        super().__init__()
        self.init_ui()

    def init_ui(self):
        self.setWindowTitle("PyQt6 Jupyter Demo")
        self.setGeometry(100, 100, 300, 200)

        layout = QVBoxLayout()

        self.label = QLabel("Klicke auf den Knopf!", self)
        layout.addWidget(self.label)

        self.button = QPushButton("Klick mich!", self)
        self.button.clicked.connect(self.on_button_click)
        layout.addWidget(self.button)

        self.setLayout(layout)

    @asyncSlot()
    async def on_button_click(self):
        self.label.setText("Bitte warte 3 Sekunden...")
        await asyncio.sleep(3)  # Asynchrone Pause
        self.label.setText("Fertig!")

# Funktion, um die Anwendung in einem Jupyter-Notebook zu starten
def start_app():
    app = QApplication.instance() or QApplication(sys.argv)
    loop = QEventLoop(app)
    asyncio.set_event_loop(loop)

    window = SimpleApp()
    window.show()

    with loop:  # Eventloop starten
        loop.run_forever()

# App starten
start_app()
```
