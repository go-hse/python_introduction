# Klassen-Eigenschaften in Python
In Python werden Klassen-Eigenschaften oft verwendet, um den Zugriff auf Attribute zu kontrollieren. Dies erfolgt üblicherweise mittels spezieller Methoden, die als `getter`, `setter` und `deleter` bekannt sind. Diese Methoden erlauben eine kontrollierte Manipulation der Eigenschaften.

## Definition von Klassen-Eigenschaften
\nUm eine Klassen-Eigenschaft zu definieren, nutzen wir den `@property`-Dekorator. Durch den Einsatz von `@property` können wir eine Methode so definieren, dass sie wie ein Attribut behandelt wird.

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius  # geschuetztes Attribut

    @property
    def radius(self):
        """Gibt den Radius zurueck."""
        return self._radius

    @radius.setter
    def radius(self, value):
        """Setzt den Radius. Soll positiv sein."""
        if value < 0:
            raise ValueError("Der Radius muss positiv sein!")
        self._radius = value
```

In diesem Beispiel wird der Radius eines Kreises als geschuetztes Attribut `_radius` gespeichert. Die Eigenschaft `radius` ist durch 
- den `@property`-Dekoator als Lese-Zugriff,
- den `@radius.setter` als Schreib-Zugriff definiert.

## Verwendung von Getter und Setter
Der `getter` wird als normale Methode ohne spezielle Parameter definiert und gibt den aktuellen Wert des Attributs zurueck. 
Der `setter` ermöglicht es, den Wert eines Attributs zu ändern, und kann Bedingungen für das Setzen des Werts festlegen. Im obigen Beispiel erzwingt der `setter`, dass der Radius eines Kreises positiv sein muss. Wird ein negativer Wert übergeben, wird ein Fehler geworfen.

```python
# Beispielhafte Verwendung
c = Circle(5)
print(c.radius)  # Ausgabe: 5

# Neuer Radius
c.radius = 10
print(c.radius)  # Ausgabe: 10

# Fehlermeldung bei falschem Wert
try:
    c.radius = -3
except ValueError as e:
    print(e)  # Ausgabe: Der Radius muss positiv sein!
```

## Weitere Funktionen
Zusätzlich zu `getter` und `setter` gibt es auch `deleter`, mit dem man das Löschen einer Eigenschaft kontrollieren kann. Allerdings werden diese weniger häufig verwendet.

Durch die Verwendung von Eigenschaften in Python können wir sicherstellen, dass die Daten unserer Objekte immer konsistent und gültig sind. Außerdem ermöglichen sie eine API, die leicht zu verstehen und zu nutzen ist.
