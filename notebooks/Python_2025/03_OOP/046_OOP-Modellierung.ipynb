## OOP Modellierung: Vererbung vs. Komposition

In der objektorientierten Programmierung (OOP) gibt es verschiedene Möglichkeiten, Beziehungen zwischen Klassen zu gestalten. Zwei wichtige Konzepte sind Vererbung und Komposition. Beide dienen dazu, Code wiederzuverwenden und komplexe Systeme zu modellieren, unterscheiden sich jedoch grundlegend in ihrer Herangehensweise und ihren Auswirkungen auf die Architektur.

Dieser Notebook erklärt die Unterschiede zwischen Vererbung und Komposition, illustriert mit Beispielen.


### Vererbung (Inheritance)

**Definition:** Vererbung ist ein Mechanismus, bei dem eine neue Klasse (die abgeleitete Klasse oder Subklasse) Eigenschaften und Methoden von einer bestehenden Klasse (der Basisklasse oder Superklasse) erbt. Die Subklasse erweitert oder modifiziert das Verhalten der Superklasse.

**`Ist-Ein` Beziehung:** Vererbung modelliert typischerweise eine "Ist-Ein"-Beziehung (is-a relationship). Zum Beispiel ist ein `Hund` ein `Säugetier`.

**Vorteile:**
*   Code-Wiederverwendung: Gemeinsame Funktionalität muss nicht neu implementiert werden.
*   Hierarchische Struktur: Modelliert natürliche Hierarchien.
*   Polymorphismus: Ermöglicht die Behandlung von Objekten verschiedener Klassen auf einheitliche Weise.

**Nachteile:**
*   Starre Struktur: Änderungen in der Basisklasse können sich unerwartet auf Subklassen auswirken.
*   Erhöhte Komplexität: Tiefe Vererbungshierarchien können schwer zu verstehen und zu warten sein.


```python
# Beispiel für Vererbung

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Generic animal sound")

class Dog(Animal):
    def speak(self):
        print("Woof!")

class Cat(Animal):
    def speak(self):
        print("Meow!")

# Verwendung
animal = Animal("Generic Animal")
dog = Dog("Fido")
cat = Cat("Whiskers")

animal.speak()
dog.speak()
cat.speak()
```

### Komposition (Composition)

**Definition:** Komposition ist ein Mechanismus, bei dem eine Klasse Objekte anderer Klassen als Bestandteile enthält. Die Klasse nutzt die Funktionalität der enthaltenen Objekte, ohne von ihnen abzuleiten.

**`Hat-Ein` Beziehung:** Komposition modelliert typischerweise eine "Hat-Ein"-Beziehung (has-a relationship). Zum Beispiel hat ein `Auto` einen `Motor`.

**Vorteile:**
*   Flexibilität: Ermöglicht das Zusammenstellen komplexer Objekte aus verschiedenen Komponenten.
*   Geringe Kopplung: Änderungen in einer Komponente wirken sich weniger auf andere Komponenten aus.
*   Bessere Testbarkeit: Einzelne Komponenten können unabhängig getestet werden.

**Nachteile:**
*   Mehr Boilerplate-Code: Die Funktionalität der Komponenten muss explizit in der enthaltenden Klasse aufgerufen werden.


```python
# Beispiel für Komposition

class Engine:
    def start(self):
        print("Engine started")

class Wheels:
    def rotate(self):
        print("Wheels are rotating")

class Car:
    def __init__(self):
        self.engine = Engine()
        self.wheels = Wheels()

    def drive(self):
        self.engine.start()
        self.wheels.rotate()

# Verwendung
car = Car()
car.drive()
```

### Unterschiede Zusammengefasst

| Merkmal          | Vererbung                                  | Komposition                                |
| ---------------- | ------------------------------------------ | ------------------------------------------ |
| Beziehung        | `Ist-Ein` (Is-A)                           | `Hat-Ein` (Has-A)                          |
| Flexibilität     | Weniger flexibel                           | Flexibler                                  |
| Kopplung         | Starke Kopplung                            | Geringe Kopplung                           |
| Code-Wiederverwendung | Durch automatische Übernahme             | Durch explizites Aufrufen der Komponenten  |
| Komplexität      | Kann zu komplexen Hierarchien führen      | Kann zu mehr Boilerplate-Code führen       |


### Wann Was Verwenden?

*   **Vererbung:** Verwenden Sie Vererbung, wenn eine klare "Ist-Ein"-Beziehung besteht und die Basisklasse eine stabile, allgemeine Schnittstelle bietet, die von allen Subklassen geteilt wird. Seien Sie vorsichtig bei tiefen Vererbungshierarchien.
*   **Komposition:** Verwenden Sie Komposition, wenn Sie Flexibilität und geringe Kopplung benötigen. Dies ist oft der Fall, wenn Objekte aus verschiedenen, unabhängigen Komponenten zusammengesetzt werden oder wenn sich die Anforderungen im Laufe der Zeit ändern können.  Bevorzugen Sie Komposition gegenüber Vererbung, um eine höhere Flexibilität zu erreichen. (Komposition over inheritance)


### Fazit

Vererbung und Komposition sind wichtige Werkzeuge in der OOP.  Komposition bietet oft mehr Flexibilität und reduziert die Kopplung zwischen Klassen.  Es ist wichtig, die Vor- und Nachteile beider Ansätze zu verstehen, um die richtige Wahl für das jeweilige Problem zu treffen.
