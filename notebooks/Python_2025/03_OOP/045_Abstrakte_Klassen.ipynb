# Abstrakte Klassen in Python

Abstrakte Klassen sind Klassen, die nicht instanziiert werden können und oftmals als Blaupausen für andere Klassen dienen. Sie enthalten eine oder mehrere abstrakte Methoden, die in den abgeleiteten Klassen implementiert werden müssen. Diese Klassen nutzen typischerweise das Modul `abc` (Abstract Base Class) in Python.

## Verwendung von Abstrakten Klassen

Abstrakte Klassen werden in Python mit Hilfe des `abc`-Moduls erstellt. Die Klasse muss von `ABC` erben, und jede Methode, die als abstrakt markiert wird, muss mit dem Dekorator `@abstractmethod` versehen sein. Hier ist ein Beispiel:

```python
from abc import ABC, abstractmethod

class Fahrzeug(ABC):
    @abstractmethod
    def antrieb(self):
        pass

class Auto(Fahrzeug):
    def antrieb(self):
        return 'Der Antrieb ist ein Verbrennungsmotor'

class Fahrrad(Fahrzeug):
    def antrieb(self):
        return 'Der Antrieb ist menschliche Kraft'
```

In diesem Beispiel ist `Fahrzeug` eine abstrakte Klasse, die eine abstrakte Methode `antrieb` definiert. Sowohl `Auto` als auch `Fahrrad` müssen die Methode `antrieb` implementieren. Versucht man, eine Instanz der abstrakten Klasse `Fahrzeug` zu erstellen, führt dies zu einem Fehler.

## Polymorphie

Polymorphie ermöglicht es, die gleiche Schnittstelle für unterschiedliche Datentypen zu verwenden. Im Beispiel oben haben wir die Methode `antrieb`, die in beiden abgeleiteten Klassen `Auto` und `Fahrrad` unterschiedlich implementiert wird. Dennoch kann auf beide Klassen auf dieselbe Weise zugegriffen werden:

```python
def print_antrieb(fahrzeug):
    print(fahrzeug.antrieb())

fahrzeuge = [Auto(), Fahrrad()]
for fahrzeug in fahrzeuge:
    print_antrieb(fahrzeug)
```

In diesem Beispiel sehen wir Polymorphie in Aktion. Die Funktion `print_antrieb` nimmt ein `Fahrzeug` als Parameter. Da sowohl `Auto` als auch `Fahrrad` die Methode `antrieb` implementieren, können wir sie in einer Liste speichern und die gleiche Funktion `print_antrieb` für beide Typen ohne Änderung des Codes aufrufen.

## Speicherung unterschiedlicher Objekt-Typen in Listen

Abstrakte Klassen und Polymorphie ermöglichen es, unterschiedliche Objekte, die von einer gemeinsamen Basisklasse abgeleitet sind, in einer einzigen Liste zu speichern. Dies wird durch die gemeinsame Schnittstelle gewährleistet, die von der abstrakten Klasse bereitgestellt wird, wie im obigen Beispiel demonstriert.

Dies erleichtert die Verwaltung und Erweiterung von Programmen, da neue Klassen hinzugefügt werden können, die bestehende abstrakte Methoden implementieren, und nahtlos in bestehende Strukturen integriert werden können, ohne bestehenden Code ändern zu müssen.
