## Typ-Hinweise in Python

Typ-Hinweise (Type Hints) sind eine Möglichkeit, den Typ einer Variablen, eines Funktionsparameters oder eines Rückgabewerts in Python anzugeben. Sie dienen der besseren Lesbarkeit und Wartbarkeit des Codes und ermöglichen es Analysewerkzeugen wie MyPy, Typfehler frühzeitig zu erkennen.


### Grundlegende Typen

Wir beginnen mit den grundlegenden Typen wie `int`, `float`, `str` und `bool`.


```python
# Type hints for basic types
age: int = 30
price: float = 99.99
name: str = "Alice"
is_active: bool = True

def greet(name: str) -> str:
    return f"Hello, {name}!"

print(greet(name))
print(f"Age: {age}, Price: {price}, Active: {is_active}")
```

### Listen

Für Listen verwenden wir `List` aus dem `typing` Modul.  Wir können den Typ der Elemente in der Liste angeben.


```python
from typing import List

numbers: List[int] = [1, 2, 3, 4, 5]
names: List[str] = ["Alice", "Bob", "Charlie"]

def process_numbers(numbers: List[int]) -> int:
    return sum(numbers)

print(f"Sum of numbers: {process_numbers(numbers)}")
```

### Dictionaries

Für Dictionaries verwenden wir `Dict` aus dem `typing` Modul. Wir geben den Typ der Schlüssel und Werte an.


```python
from typing import Dict

person: Dict[str, str] = {"name": "Alice", "city": "Berlin"}
ages: Dict[str, int] = {"Alice": 30, "Bob": 25}

def get_age(ages: Dict[str, int], name: str) -> int:
    return ages[name]

print(f"Alice's age: {get_age(ages, 'Alice')}")
```

### Any

Wenn der Typ einer Variablen oder eines Rückgabewerts unbekannt oder beliebig sein kann, verwenden wir `Any` aus dem `typing` Modul.


```python
from typing import Any

data: Any = 123
data = "Hello"
data = [1, 2, 3]

def process_data(data: Any) -> Any:
    return data

print(process_data(data))
```

### Klassen

Typ-Hinweise können auch in Klassen verwendet werden, um die Typen von Attributen und Methodenparametern anzugeben.


```python
class Person:
    name: str
    age: int

    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def greet(self) -> str:
        return f"Hello, my name is {self.name} and I am {self.age} years old."

person = Person("Alice", 30)
print(person.greet())
```
