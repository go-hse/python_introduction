# Objekt-Orientierte Programmierung in Python
Objekt-orientierte Programmierung (OOP) ist ein Programmierparadigma, das auf dem Konzept von "Objekten" basiert, die Daten in Form von Feldern (auch bekannt als Attribute) und Code in Form von Funktionen/Prozeduren (bekannt als Methoden) enthalten. Die drei Eckpfeiler der OOP sind __Kapselung__, __Vererbung__ und __Polymorphie__.

## Klassen und Objekte
Eine Klasse ist ein Bauplan für Objekte. Ein Objekt ist eine Instanz einer Klasse. In Python wird eine Klasse mit dem Schlüsselwort `class` definiert. Die Funktionen (def) in Klassen heißen Methoden und definieren das __Verhalten__ der Objekte. 

```python
class Auto:
    def __init__(self, marke, modell):
        self.marke = marke
        self.modell = modell

mein_auto = Auto("VW", "Golf")
print(mein_auto.marke)
```

In diesem Beispiel wird eine Klasse `Auto` erstellt, und `mein_auto` ist eine Instanz dieser Klasse.

## Kapselung - Zugriffsmechanismen
In Python gibt es drei Hauptzugriffsmechanismen für Attribute und Methoden: öffentlich, geschützt und privat.

- **Öffentlich**: Diese sind für alle sichtbar (Standard in Python).
- **Geschützt**: Konventionell mit einem Unterstrich `_` markiert. Sollte nicht außerhalb der Klasse verändert werden.
- **Privat**: Durch Doppelt-Unterstrich `__` gekennzeichnet. Wird durch Namensveränderung eingeschränkt.

```python
class Test:
    def __init__(self):
        self.oeffentlich = 'Ich bin öffentlich'
        self._geschützt = 'Ich bin geschützt'
        self.__privat = 'Ich bin privat'

obj = Test()
print(obj.oeffentlich)  # Zugriff möglich
print(obj._geschützt)  # Zugriff möglich, aber nicht empfohlen
# print(obj.__privat)  # Führt zu einem Fehler
```

Für ein privates Attribut kann man dennoch über `obj._Klasse__attribut` zugreifen:

````python
class Test:
    def __init__(self):
        self.oeffentlich = 'Ich bin öffentlich'
        self._geschuetzt = 'Ich bin geschützt'
        self.__privat = 'Ich bin privat'

obj = Test()
print(obj._Test__privat)
print(obj.__privat)
````

## Vererbung
Vererbung ermöglicht es einer Klasse, die Eigenschaften und Methoden einer anderen Klasse zu erben.

```python
class Fahrzeug:
    def __init__(self, farbe):
        self.farbe = farbe

    def starten(self):
        print("Das Fahrzeug startet")

class Auto(Fahrzeug):
    def __init__(self, farbe, marke):
        super().__init__(farbe)
        self.marke = marke

mein_auto = Auto("Blau", "VW")
print(mein_auto.farbe)
mein_auto.starten()
```

In diesem Beispiel erbt die Klasse `Auto` die Attribute und Methoden der Klasse `Fahrzeug`. `Auto` ist eine Unterklasse von `Fahrzeug`.

## Polymorphie
Polymorphie erlaubt es, auf dieselbe Art auf verschiedene Objekttypen zuzugreifen. Es bedeutet, dass eine Methode in unterschiedlichen Klassen unterschiedliche Implementierungen haben kann.

```python
class Katze:
    def sprache(self):
        print("Miau")

class Hund:
    def sprache(self):
        print("Wuff")

# Funktion, die polymorph arbeitet
def tier_sprache(tier):
    tier.sprache()

katze = Katze()
hund = Hund()

tier_sprache(katze)
# Ausgabe: Miau
tier_sprache(hund)
# Ausgabe: Wuff
```

Hier zeigt sich Polymorphie in der Fähigkeit der Funktion `tier_sprache`, Objekte sowohl der Klasse `Katze` als auch der Klasse `Hund` zu akzeptieren und die jeweilige `sprache`-Methode aufzurufen.

## Statische Attribute und Methoden
Statische Attribute und Methoden sind an die Klasse selbst gebunden und nicht an die Instanz der Klasse.

### Statische Attribute
Statische Attribute werden in Python oft außerhalb jeder Methode und mit einem Klassennamen referenziert.

```python
class Mathematik:
    pi = 3.14159265359  # Statisches Attribut

print(Mathematik.pi)
```

### Statische Methoden
Statische Methoden werden mit dem Dekorator `@staticmethod` deklariert.

```python
class Mathematik:

    @staticmethod
    def addiere(x, y):
        return x + y

print(Mathematik.addiere(5, 7))
```

In diesem Beispiel ist `addiere` eine statische Methode, die nicht auf Instanzdaten zugreift, und `pi` ist ein statisches Attribut.

__Statische Methoden__ werden an abgeleitete Klassen __vererbt__.
