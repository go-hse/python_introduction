# Spezielle Methoden (Magic Methods) in Python
Python-Klassen bieten spezielle Methoden, auch als Magic Methods oder Dunder-Methoden (Double Underscore) bezeichnet, die es ermöglichen, die Standardfunktionen von Objekten zu erweitern. Diese Methoden haben spezielle Namen, die mit doppeltem Unterstrich beginnen und enden, wie z. B. `__init__`, `__str__` und `__repr__`.


1. Initialisierung und Repräsentation
```
    __init__(self, ...): Konstruktor, initialisiert ein Objekt.
    __del__(self): Destruktor, wird aufgerufen, wenn ein Objekt gelöscht wird.
    __repr__(self): Formale String-Repräsentation eines Objekts, oft für Debugging.
    __str__(self): Informelle String-Repräsentation, z. B. für print().
    __bytes__(self): Gibt die Byte-Repräsentation eines Objekts zurück.
    __format__(self, format_spec): Anpassung der String-Ausgabe mit Formatangaben.
```

2. Vergleichsoperatoren

```
    __eq__(self, other): Für == (Gleichheit).
    __ne__(self, other): Für != (Ungleichheit).
    __lt__(self, other): Für < (Kleiner als).
    __le__(self, other): Für <= (Kleiner oder gleich).
    __gt__(self, other): Für > (Größer als).
    __ge__(self, other): Für >= (Größer oder gleich).
```

3. Arithmetische Operatoren

    Addition und Subtraktion:
```
        __add__(self, other): Für +.
        __sub__(self, other): Für -.
```
    Multiplikation und Division:
```
        __mul__(self, other): Für *.
        __truediv__(self, other): Für /.
        __floordiv__(self, other): Für //.
        __mod__(self, other): Für %.
```
    Potenzierung:
```
        __pow__(self, other[, modulo]): Für **.
```
    Bitweise Operationen:
```
        __and__(self, other): Für &.
        __or__(self, other): Für |.
        __xor__(self, other): Für ^.
        __lshift__(self, other): Für <<.
        __rshift__(self, other): Für >>.
```

4. Reflektierte arithmetische Operatoren

Wenn das linke Objekt den Operator nicht unterstützt:

```
    __radd__(self, other): Für +.
    __rsub__(self, other): Für -.
    __rmul__(self, other): Für *, usw.
```

5. In-Place Operatoren

Für Änderungen direkt im Objekt:

```
    __iadd__(self, other): Für +=.
    __isub__(self, other): Für -=.
    __imul__(self, other): Für *=, usw.
```

6. Container-Schnittstellen

```
    __len__(self): Länge eines Containers (len()).
    __getitem__(self, key): Zugriff auf Elemente (obj[key]).
    __setitem__(self, key, value): Zuweisung von Werten (obj[key] = value).
    __delitem__(self, key): Löschen von Werten (del obj[key]).
    __iter__(self): Iteration (for-Schleifen).
    __next__(self): Nächster Wert in einer Iteration.
    __contains__(self, item): Für in (Mitgliedschaftsprüfung).
```

7. Kontextmanager

```
    __enter__(self): Einstieg in einen Kontextblock (with-Anweisung).
    __exit__(self, exc_type, exc_value, traceback): Austritt aus einem Kontextblock.
```

8. Attributzugriff

```
    __getattr__(self, name): Zugriff auf nicht vorhandene Attribute.
    __getattribute__(self, name): Zugriff auf alle Attribute.
    __setattr__(self, name, value): Zuweisung eines Attributwerts.
    __delattr__(self, name): Löschen eines Attributs.
    __dir__(self): Für die Auflistung der Attribute (dir()).
```

9. Metaprogrammierung

```
    __call__(self, ...): Macht ein Objekt aufrufbar wie eine Funktion.
    __instancecheck__(self, instance): Prüfung für isinstance().
    __subclasscheck__(self, subclass): Prüfung für issubclass().
```

10. Sonstige Methoden

```
    __hash__(self): Hash-Wert eines Objekts (hash()).
    __bool__(self): Boolescher Wert eines Objekts (bool()).
    __copy__(self): Shallow Copy eines Objekts.
    __deepcopy__(self, memo): Deep Copy eines Objekts.
```

    

## 1. `__init__`
Die `__init__`-Methode wird verwendet, um ein Objekt zu initialisieren, wenn eine neue Instanz der Klasse erstellt wird.

````python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

# Beispiel
person = Person("Alice", 30)
print(person.name)  # Ausgabe: Alice
print(person.alter) # Ausgabe: 30
````

## 2. `__str__`
Die `__str__`-Methode definiert, wie ein Objekt als String dargestellt wird, z. B. wenn `print(objekt)` aufgerufen wird.

````python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def __str__(self):
        return f"{self.name}, {self.alter} Jahre alt"

# Beispiel
person = Person("Alice", 30)
print(person)  # Ausgabe: Alice, 30 Jahre alt
````

## 3. `__repr__`
Die `__repr__`-Methode gibt eine formale Darstellung eines Objekts zurück, die oft zur Debugging-Zwecken verwendet wird. Ziel ist es, dass der zurückgegebene String zur Wiedererstellung des Objekts genutzt werden kann.

````python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    def __repr__(self):
        return f"Person(\"{self.name}\", {self.alter})"

# Beispiel
person = Person("Alice", 30)
print(repr(person))  # Ausgabe: Person("Alice", 30)
````

## 4. `__add__`
Die `__add__`-Methode definiert das Verhalten des Objekts für den `+`-Operator.

````python
class Vektor:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, anderer):
        return Vektor(self.x + anderer.x, self.y + anderer.y)

    def __repr__(self):
        return f"Vektor({self.x}, {self.y})"

# Beispiel
v1 = Vektor(2, 3)
v2 = Vektor(1, 4)
print(v1 + v2)  # Ausgabe: Vektor(3, 7)
````

## 5. `__len__`
Die `__len__`-Methode definiert, was passiert, wenn die eingebaute Funktion `len()` auf das Objekt angewendet wird.

````python
class Sammlung:
    def __init__(self, elemente):
        self.elemente = elemente

    def __len__(self):
        return len(self.elemente)

# Beispiel
sammlung = Sammlung([1, 2, 3, 4])
print(len(sammlung))  # Ausgabe: 4
````

## Fazit
Spezielle Methoden ermöglichen es, Klassen intuitiver und Python-typischer zu gestalten. Sie machen den Code einfacher zu lesen und zu verwenden, indem sie die Standardoperationen auf benutzerdefinierte Objekte anwenden.
